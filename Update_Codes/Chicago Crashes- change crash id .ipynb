```python
import pandas as pd 
dfv = pd.read_csv('Traffic_Crashes_Vehicles_20240409.csv', low_memory=False)
dfd = pd.read_csv('Traffic_Crashes_People_20240409.csv', low_memory=False)
```

```python
all_old_ids = set(pd.unique(dfv['CRASH_RECORD_ID']).tolist() + pd.unique(dfd['CRASH_RECORD_ID']).tolist())
new_id_mapping = {old_id: new_id for new_id, old_id in enumerate(all_old_ids)}

dfd['new_id'] = dfd['CRASH_RECORD_ID'].map(new_id_mapping)
dfv['new_id'] = dfv['CRASH_RECORD_ID'].map(new_id_mapping)
```

```python
dfd = dfd.drop(columns=['CRASH_RECORD_ID'])
dfv = dfv.drop(columns=['CRASH_RECORD_ID'])
```

```python
dfd.rename(columns={'new_id': 'CRASH_RECORD_ID'}, inplace=True)
dfv.rename(columns={'new_id': 'CRASH_RECORD_ID'}, inplace=True)
```
